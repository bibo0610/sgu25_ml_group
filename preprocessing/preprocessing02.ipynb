## 3. Lọc các cột dữ liệu lệch mạnh 

```python
del_col = ['Id', 'BsmtFinSF2', 'LowQualFinSF', 'BsmtFullBath', 'BsmtHalfBath', 'EnclosedPorch', '3SsnPorch', 'ScreenPorch', 'PoolArea', 'MiscVal', 'Street', 'Utilities', 'LandSlope', 'Condition1', 'Condition2', 'BldgType', 'RoofMatl', 'ExterCond', 'BsmtCond', 'BsmtFinType2', 'Heating', 'Electrical', 'Functional', 'GarageQual', 'GarageCond', 'PavedDrive', 'MiscFeature', 'SaleType', 'SaleCondition']

def dropCol(df):
    for name in del_col:
        df.drop(name,axis=1,inplace=True)
    return df
```

## 4. Kiểm tra giá trị thiếu 

```python
# Kiểm tra giá trị thiếu
def isNull(df):
    print("="*5, "Checking Null", "="*5)
    null_counts = df.isnull().sum()
    null_cols = null_counts[null_counts > 0]
    if len(null_cols) > 0 :
        display.display(null_cols)
    has_null = df.isnull().sum().any()
    print(f'--> Có giá trị Null: {has_null}')
```
