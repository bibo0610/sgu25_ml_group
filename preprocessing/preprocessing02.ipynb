```python
def report_missing(df, name="df"):
    miss = df.isnull().sum()
    miss = miss[miss > 0].sort_values(ascending=False)
    print(f"\nMissing in {name}: {miss.shape[0]} columns")
    if miss.shape[0] > 0:
        print(miss)
    else:
        print("No missing values")

report_missing(df_train, "df_train (before)")
report_missing(df_test, "df_test (before)")
```

### Xử lý Missing Values

```python
cols_train = set(df_train.columns)
cols_test  = set(df_test.columns)

# --------------------
# Special-case fills
# --------------------

# 1) LotFrontage: fill by median LotFrontage per Neighborhood (if LotFrontage present)
if 'LotFrontage' in cols_train and 'Neighborhood' in cols_train:
    neigh_median = df_train.groupby('Neighborhood')['LotFrontage'].median()
    df_train['LotFrontage'] = df_train['LotFrontage'].fillna(df_train['Neighborhood'].map(neigh_median))
    if 'LotFrontage' in cols_test and 'Neighborhood' in cols_test:
        df_test['LotFrontage'] = df_test['LotFrontage'].fillna(df_test['Neighborhood'].map(neigh_median))
    # remaining NaN will be handled by median imputer below

# 2) MasVnrArea: you have MasVnrArea but may not have MasVnrType.
# If MasVnrType exists, set area to 0 where type == 'None'; otherwise as a safe choice fill 0.
if 'MasVnrArea' in cols_train:
    if 'MasVnrType' in cols_train:
        # if type explicitly 'None', ensure area zero
        df_train.loc[df_train['MasVnrType'].fillna('None') == 'None', 'MasVnrArea'] = df_train.loc[
            df_train['MasVnrType'].fillna('None') == 'None', 'MasVnrArea'].fillna(0)
        if 'MasVnrType' in cols_test and 'MasVnrArea' in cols_test:
            df_test.loc[df_test['MasVnrType'].fillna('None') == 'None', 'MasVnrArea'] = df_test.loc[
                df_test['MasVnrType'].fillna('None') == 'None', 'MasVnrArea'].fillna(0)
    else:
        # No MasVnrType column — safer to assume missing -> 0 (no veneer)
        df_train['MasVnrArea'] = df_train['MasVnrArea'].fillna(0)
        if 'MasVnrArea' in cols_test:
            df_test['MasVnrArea'] = df_test['MasVnrArea'].fillna(0)

# 3) FireplaceQu: treat NaN as "None" (no fireplace quality)
if 'FireplaceQu' in cols_train:
    df_train['FireplaceQu'] = df_train['FireplaceQu'].fillna('None')
    if 'FireplaceQu' in cols_test:
        df_test['FireplaceQu'] = df_test['FireplaceQu'].fillna('None')

# 4) GarageYrBlt: if missing, try to fill with YearRemodAdd (if exists), else YearBuilt if exists, else median, else 0
if 'GarageYrBlt' in cols_train:
    if 'YearRemodAdd' in cols_train:
        df_train['GarageYrBlt'] = df_train['GarageYrBlt'].fillna(df_train['YearRemodAdd'])
    elif 'YearBuilt' in cols_train:
        df_train['GarageYrBlt'] = df_train['GarageYrBlt'].fillna(df_train['YearBuilt'])
    # leftover NaN -> will be imputed by median below; but to be safe set to median now
    if df_train['GarageYrBlt'].isnull().any():
        df_train['GarageYrBlt'] = df_train['GarageYrBlt'].fillna(df_train['GarageYrBlt'].median())

    if 'GarageYrBlt' in cols_test:
        if 'YearRemodAdd' in cols_test:
            df_test['GarageYrBlt'] = df_test['GarageYrBlt'].fillna(df_test['YearRemodAdd'])
        elif 'YearBuilt' in cols_test:
            df_test['GarageYrBlt'] = df_test['GarageYrBlt'].fillna(df_test['YearBuilt'])
        # remaining NaN -> fill with train median (if exists) else 0
        train_med = df_train['GarageYrBlt'].median() if 'GarageYrBlt' in df_train.columns else 0
        df_test['GarageYrBlt'] = df_test['GarageYrBlt'].fillna(train_med)

# 5) If any of categorical "Qual/Cond/Finish/Type/Bsmt" exist and have NaN, replace with 'None' (meaning missing)
qual_like = [c for c in df_train.columns if c.lower().find('qual') != -1 or c.lower().find('cond') != -1 or
             c.lower().find('finish') != -1 or c.lower().find('type') != -1 or c.lower().find('bsmt') != -1 or
             c.lower().find('exter') != -1 or c.lower().find('kitchen') != -1]
# limit to object dtypes
for c in qual_like:
    if c in df_train.columns and df_train[c].dtype == 'O':
        df_train[c] = df_train[c].fillna('None')
        if c in df_test.columns:
            df_test[c] = df_test[c].fillna('None')

# --------------------
# General imputation
# --------------------

# Numeric columns in train
num_cols_train = df_train.select_dtypes(include=[np.number]).columns.tolist()
if 'SalePrice' in num_cols_train:
    num_cols_train.remove('SalePrice')
# Categorical/object columns in train
cat_cols_train = df_train.select_dtypes(include=['object']).columns.tolist()

# Create imputers
num_imputer = SimpleImputer(strategy='median')
cat_imputer = SimpleImputer(strategy='most_frequent')

# Fit imputers on train (if columns exist)
if len(num_cols_train) > 0:
    num_imputer.fit(df_train[num_cols_train])
if len(cat_cols_train) > 0:
    cat_imputer.fit(df_train[cat_cols_train])

# Transform train
if len(num_cols_train) > 0:
    df_train[num_cols_train] = num_imputer.transform(df_train[num_cols_train])
if len(cat_cols_train) > 0:
    df_train[cat_cols_train] = cat_imputer.transform(df_train[cat_cols_train])

# Transform test for the intersection of columns only
if len(num_cols_train) > 0:
    num_common = [c for c in num_cols_train if c in df_test.columns]
    if num_common:
        df_test[num_common] = num_imputer.transform(df_test[num_common])

if len(cat_cols_train) > 0:
    cat_common = [c for c in cat_cols_train if c in df_test.columns]
    if cat_common:
        df_test[cat_common] = cat_imputer.transform(df_test[cat_common])

# Final check
report_missing(df_train, "df_train (after)")
report_missing(df_test, "df_test (after)")
```

**Nhận xét**
- LotFrontage được điền bằng median theo từng Neighborhood, hợp lý vì mặt tiền phụ thuộc khu vực.

- MasVnrArea được điền 0 khi MasVnrType là “None”, phản ánh đúng việc không có ốp đá.

- FireplaceQu thiếu được gán “None”, nghĩa là nhà không có lò sưởi.

- GarageYrBlt được điền bằng YearRemodAdd, hoặc YearBuilt, còn lại thì dùng median.

- Các cột liên quan chất lượng/điều kiện (qual, cond, bsmt, exter, kitchen…) thiếu được đưa về “None”, thể hiện không có hạng mục.

Sau khi xử lý các trường hợp đặc biệt, phần còn lại dùng median cho số và most_frequent cho phân loại để đảm bảo train–test đồng nhất và không rò rỉ dữ liệu.

### Label Encoding cho các cột phân loại

```python
from sklearn.preprocessing import LabelEncoder
# Lưu lại cột target
y = df_train['SalePrice']

# Xóa SalePrice khỏi train trước khi encoding
X_train = df_train.drop(columns=['SalePrice'])
X_test  = df_test.copy()

categorical_cols = X_train.select_dtypes(include=['object']).columns

# Tạo dict lưu encoder cho mỗi cột (để encode test giống train)
encoders = {}

# ---- label encoding ----
for col in categorical_cols:
    le = LabelEncoder()
    
    # Fit trên train
    X_train[col] = le.fit_transform(X_train[col].astype(str))
    
    # Transform trên test, với các giá trị mới -> gán -1
    X_test[col] = X_test[col].astype(str).map(
        lambda x: le.transform([x])[0] if x in le.classes_ else -1
    )

    encoders[col] = le

# Ghép lại target vào train
df_train_encoded = X_train.copy()
df_train_encoded['SalePrice'] = y

df_test_encoded = X_test.copy()

# Kiểm tra kích thước
print("Train encoded:", df_train_encoded.shape)
print("Test encoded:", df_test_encoded.shape)
```

### Chuẩn hóa (Scaling) các biến số liên tục

```python
# Tách target
y = df_train_encoded['SalePrice']
X_train_encoded = df_train_encoded.drop(columns=['SalePrice']).copy()
X_test_encoded  = df_test_encoded.copy()

# Tìm các cột numeric trong train
num_cols = X_train_encoded.select_dtypes(include=[np.number]).columns.tolist()

# Loại các cột nhị phân (one-hot) khỏi danh sách scale:
# định nghĩa nhị phân nếu chỉ có 2 giá trị duy nhất (ví dụ 0/1)
binary_cols = [c for c in num_cols if X_train_encoded[c].nunique() <= 2]

# Các cột sẽ scale = num_cols - binary_cols
cols_to_scale = [c for c in num_cols if c not in binary_cols]

print(f"Tổng numeric columns: {len(num_cols)}")
print(f"Binary (không scale): {len(binary_cols)} -> sample: {binary_cols[:10]}")
print(f"Columns to scale (continuous): {len(cols_to_scale)} -> sample: {cols_to_scale[:10]}")

# Nếu không có cột nào để scale thì bỏ qua
if len(cols_to_scale) > 0:
    scaler = StandardScaler()
    # Fit trên train
    scaler.fit(X_train_encoded[cols_to_scale])

    # Transform cả train & test (chỉ những cột có trong test)
    X_train_encoded[cols_to_scale] = scaler.transform(X_train_encoded[cols_to_scale])

    # đảm bảo test có các cột cần thiết
    common_scale_cols = [c for c in cols_to_scale if c in X_test_encoded.columns]
    if len(common_scale_cols) > 0:
        X_test_encoded[common_scale_cols] = scaler.transform(X_test_encoded[common_scale_cols])

print("\nScaling finished.")
print("X_train shape:", X_train_encoded.shape)
print("X_test shape :", X_test_encoded.shape)

# Ghép lại target để dễ dùng tiếp
df_train_scaled = X_train_encoded.copy()
df_train_scaled['SalePrice'] = y
df_test_scaled = X_test_encoded.copy()

# Kiểm tra nhanh: thống kê các cột đã scale
if len(cols_to_scale) > 0:
    display.display(df_train_scaled[cols_to_scale].describe().loc[['mean','std']])
else:
    print("No continuous numeric columns to scale.")
```
