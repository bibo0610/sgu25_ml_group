```python
def report_missing(df, name="df"):
    miss = df.isnull().sum()
    miss = miss[miss > 0].sort_values(ascending=False)
    print(f"\nMissing in {name}: {miss.shape[0]} columns")
    if miss.shape[0] > 0:
        print(miss)
    else:
        print("No missing values")

report_missing(df_train, "df_train (before)")
report_missing(df_test, "df_test (before)")
```

### Xử lý Missing Values

```python
# ===============================
# 1) Missing = "None" (không có)
# ===============================
none_cols = [
    "Alley","BsmtQual","BsmtCond","BsmtExposure","BsmtFinType1","BsmtFinType2",
    "FireplaceQu","GarageType","GarageFinish","GarageQual","GarageCond",
    "MasVnrType","PoolQC","Fence","MiscFeature"
]

for col in none_cols:
    if col in df_train.columns:
        df_train[col] = df_train[col].fillna("None")
    if col in df_test.columns:
        df_test[col] = df_test[col].fillna("None")

# ===============================
# 2) Missing = 0 (không có diện tích / số lượng)
# ===============================
zero_cols = [
    "MasVnrArea","BsmtFinSF1","BsmtFinSF2","BsmtUnfSF","TotalBsmtSF",
    "BsmtFullBath","BsmtHalfBath","GarageArea","GarageCars","GarageYrBlt"
]

for col in zero_cols:
    if col in df_train.columns:
        df_train[col] = df_train[col].fillna(0)
    if col in df_test.columns:
        df_test[col] = df_test[col].fillna(0)

# ===============================
# 3) LotFrontage = median theo Neighborhood
# ===============================
if "LotFrontage" in df_train.columns:
    neigh_med = df_train.groupby("Neighborhood")["LotFrontage"].median()
    df_train["LotFrontage"] = df_train["LotFrontage"].fillna(df_train["Neighborhood"].map(neigh_med))
    df_test["LotFrontage"]  = df_test["LotFrontage"].fillna(df_test["Neighborhood"].map(neigh_med))

# ===============================
# 4) Mode for categorical còn lại 
# ===============================
cat_cols = df_train.select_dtypes(include=["object"]).columns.tolist()

for col in cat_cols:
    mode_val = df_train[col].mode()[0]
    df_train[col] = df_train[col].fillna(mode_val)
    if col in df_test.columns:
        df_test[col] = df_test[col].fillna(mode_val)

# ===============================
# 5) Median for numeric còn lại
# ===============================
num_cols = df_train.select_dtypes(include=["int64","float64"]).columns.tolist()
if "SalePrice" in num_cols: 
    num_cols.remove("SalePrice")

for col in num_cols:
    med = df_train[col].median()
    df_train[col] = df_train[col].fillna(med)
    if col in df_test.columns:
        df_test[col] = df_test[col].fillna(med)

# ===============================
# Check again
# ===============================
report_missing(df_train, "df_train (after)")
report_missing(df_test, "df_test (after)")
```

```python
# --- Target ---
y = df_train["SalePrice"].copy()
X_train = df_train.drop(columns=["SalePrice"])
X_test  = df_test.copy()

# --- Tách numeric / categorical ---
num_cols = X_train.select_dtypes(include=["number"]).columns.tolist()
cat_cols = X_train.select_dtypes(exclude=["number"]).columns.tolist()

# --- Các cột ordinal (có thứ tự thật) ---
ordinal_cols = [
    "ExterQual","ExterCond","BsmtQual","BsmtCond","HeatingQC","KitchenQual",
    "FireplaceQu","GarageQual","GarageCond","PoolQC",
    "BsmtExposure","BsmtFinType1","BsmtFinType2","Functional",
    "GarageFinish","LotShape","LandSlope","PavedDrive","Utilities",
    "CentralAir"
]
ordinal_cols = [c for c in ordinal_cols if c in cat_cols]
nominal_cols = [c for c in cat_cols if c not in ordinal_cols]

# --- Thứ tự cho ordinal ---
qual_order = ["None","Po","Fa","TA","Gd","Ex"]
bsmt_exposure_order = ["None","No","Mn","Av","Gd"]
bsmt_fin_order = ["None","Unf","LwQ","Rec","BLQ","ALQ","GLQ"]
functional_order = ["Sal","Sev","Maj2","Maj1","Mod","Min2","Min1","Typ"]
garage_finish_order = ["None","Unf","RFn","Fin"]
lotshape_order = ["IR3","IR2","IR1","Reg"]
landslope_order = ["Sev","Mod","Gtl"]
paveddrive_order = ["N","P","Y"]
utilities_order = ["ELO","NoSeWa","NoSewr","AllPub"]
centralair_order = ["N","Y"]

ordinal_categories = []
for c in ordinal_cols:
    if c in ["ExterQual","ExterCond","BsmtQual","BsmtCond","HeatingQC","KitchenQual",
             "FireplaceQu","GarageQual","GarageCond","PoolQC"]:
        ordinal_categories.append(qual_order)
    elif c == "BsmtExposure":
        ordinal_categories.append(bsmt_exposure_order)
    elif c in ["BsmtFinType1","BsmtFinType2"]:
        ordinal_categories.append(bsmt_fin_order)
    elif c == "Functional":
        ordinal_categories.append(functional_order)
    elif c == "GarageFinish":
        ordinal_categories.append(garage_finish_order)
    elif c == "LotShape":
        ordinal_categories.append(lotshape_order)
    elif c == "LandSlope":
        ordinal_categories.append(landslope_order)
    elif c == "PavedDrive":
        ordinal_categories.append(paveddrive_order)
    elif c == "Utilities":
        ordinal_categories.append(utilities_order)
    elif c == "CentralAir":
        ordinal_categories.append(centralair_order)
    else:
        ordinal_categories.append(qual_order)

# --- Pipelines ---
num_pipe = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

ord_pipe = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="constant", fill_value="None")),
    ("ordinal", OrdinalEncoder(
        categories=ordinal_categories,
        handle_unknown="use_encoded_value",
        unknown_value=-1
    ))
])

nom_pipe = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", num_pipe, num_cols),
        ("ord", ord_pipe, ordinal_cols),
        ("nom", nom_pipe, nominal_cols)
    ],
    remainder="drop"
)

# --- Fit train, transform test ---
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed  = preprocessor.transform(X_test)

# --- Feature names sau onehot ---
feature_names = []
feature_names.extend(num_cols)
feature_names.extend(ordinal_cols)

ohe = preprocessor.named_transformers_["nom"].named_steps["onehot"]
feature_names.extend(ohe.get_feature_names_out(nominal_cols))

df_train_encoded = pd.DataFrame(X_train_processed, columns=feature_names)
df_test_encoded  = pd.DataFrame(X_test_processed, columns=feature_names)

print("Train encoded:", df_train_encoded.shape)
print("Test encoded :", df_test_encoded.shape)
```

```python
# ---- 1) Target lấy từ data gốc ----
data_path = "../../data02"
df_train_org = pd.read_csv(f"{data_path}/train.csv")
y = df_train_org["SalePrice"].copy()

X_train = df_train_encoded.copy()
X_test  = df_test_encoded.copy()

# ---- 2) scale continuous numeric (không scale one-hot/binary, không scale ordinal) ----
num_cols = X_train.select_dtypes(include=[np.number]).columns.tolist()

binary_cols = [c for c in num_cols if X_train[c].nunique() == 2]

ordinal_cols_set = set([
    "ExterQual","ExterCond","BsmtQual","BsmtCond","HeatingQC","KitchenQual",
    "FireplaceQu","GarageQual","GarageCond","PoolQC","BsmtExposure",
    "BsmtFinType1","BsmtFinType2","Functional","GarageFinish",
    "LotShape","LandSlope","PavedDrive","Utilities","CentralAir"
])
ordinal_cols = [c for c in X_train.columns if c in ordinal_cols_set]

cols_to_scale = [c for c in num_cols if c not in binary_cols and c not in ordinal_cols]

if cols_to_scale:
    scaler = StandardScaler()
    scaler.fit(X_train[cols_to_scale])

    X_train[cols_to_scale] = scaler.transform(X_train[cols_to_scale])
    X_test[cols_to_scale]  = scaler.transform(X_test[cols_to_scale])

print("Scaling finished.")
print("X_train shape:", X_train.shape)
print("X_test shape :", X_test.shape)

# ---- 3) gắn lại target để train ----
df_train_scaled = X_train.copy()
df_train_scaled["SalePrice"] = y.values

df_test_scaled = X_test.copy()
```
