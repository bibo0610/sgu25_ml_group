```python
# Kiểm tra giá trị thiếu
def isNull(df):
    print("="*5, "Checking Null", "="*5)
    null_counts = df.isnull().sum()
    null_cols = null_counts[null_counts > 0]
    if len(null_cols) > 0 :
        display.display(null_cols)
    has_null = df.isnull().sum().any()
    print(f'--> Có giá trị Null: {has_null}')
```

# Trực quan hóa dữ liệu

### Phân bố từng thuộc tính

```python
for cols in numeric_cols:
    plt.figure(figsize=(6,4))
    sns.histplot(x=cols, data=df_train)
    plt.title(f"Phân bố của {cols}")
    plt.show()
```

```python
for col in categorical_cols:
    plt.figure(figsize=(8, 4))
    sns.countplot(x=col, data=df_train, order=df_train[col].value_counts().index)
    plt.title(f'Phân bố của {col}', fontsize=12, fontweight='bold')
    
    # Nếu có quá nhiều giá trị thì xoay nhãn
    if df_train[col].nunique() > 5:
        plt.xticks(rotation=45, ha='right')
    
    plt.tight_layout()
    plt.show()
```

### Trực quan hóa theo biến mục tiêu

```python
# Tạo biến mục tiêu phân loại (chia SalePrice thành 3 nhóm)
df_train['PriceRange'] = pd.qcut(df_train['SalePrice'], q=3, labels=['Thấp', 'Trung bình', 'Cao'])

# Biểu đồ countplot theo biến mục tiêu
for col in categorical_cols:
    plt.figure(figsize=(10, 5))
    sns.countplot(x=col, hue='PriceRange', data=df_train)
    plt.title(f'Phân bố {col} theo nhóm giá nhà (SalePrice)')
    plt.xticks(rotation=45, ha='right')
    plt.legend(title='Nhóm giá')
    plt.tight_layout()
    plt.show()
```

### Tương quan giữa các biến số với SalePrice

```python
corr = df_train.corr(numeric_only=True)
plt.figure(figsize=(12,8))
sns.heatmap(corr, cmap='coolwarm', center=0)
plt.title('Ma trận tương quan giữa các biến số')
plt.show()

# Top 10 biến tương quan mạnh nhất với SalePrice
corr_target = corr['SalePrice'].sort_values(ascending=False)
print("Top 10 biến tương quan mạnh nhất với SalePrice:")
print(corr_target.head(10))
```

### Phát hiện giá trị ngoại lai (Outlier)

```python
top_corr_features = corr_target.index[1:6]
for col in top_corr_features:
    plt.figure(figsize=(6,4))
    sns.boxplot(x=df_train[col])
    plt.title(f"Boxplot of {col}")
    plt.show()
```

**Nhận xét**
- OverallQual có tương quan mạnh nhất với SalePrice. Khi chất lượng nhà tăng thì giá nhà cũng tăng  -> đây là biến dự báo rất quan trọng
- GrLivArea giá bán sẽ tăng theo diện tích đất sử dụng
- GarageCars và GarageArea cùng thể hiện khả năng chứa xe của ngôi nhà -> ảnh hưởng khá cao đến giá nhà
- TotalBsmtSF: tổng diện tích tầng hầm cũng có khả năng ảnh hưởng khá cao đến giá nhà vì độ tương quan khá cao (0.613581)
